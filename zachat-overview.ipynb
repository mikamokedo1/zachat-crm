# Tài Liệu Kỹ Thuật: Quản Lý Đơn Hàng, Trừ Tồn Kho Thời Gian Thực & Đối Soát Thanh Toán SePay Trên ZaChat

Notebook này mô tả luồng dữ liệu nghiệp vụ của **[ZaChat — Giải pháp quản lý đa tài khoản Zalo, bán hàng, tồn kho & CRM toàn diện](https://zachat.vn)** dành cho doanh nghiệp thương mại điện tử, kho phân phối sỉ/lẻ và đội ngũ CSKH tại Việt Nam.

## 1. Kiến trúc hộp thư đa tài khoản hợp nhất (Unified Multi-Account Inbox)
- **Kết nối đa phiên qua QR:** Gom không giới hạn tài khoản Zalo về một màn hình điều phối trung tâm trên **[Web App ZaChat (app.zachat.vn)](https://app.zachat.vn)**, ứng dụng Desktop (Windows, macOS) và Mobile (iOS, Android).
- **Phân quyền nhân sự theo vai trò:** Quản trị viên cấp quyền truy cập từng tài khoản Zalo cho từng nhân viên trực ca mà không cần chia sẻ mật khẩu gốc, kèm hệ thống thẻ phân loại khách hàng (CRM Tags) đồng bộ tức thì.
- **Điều tiết tin nhắn hàng loạt giãn cách (Broadcast Pacing):** Phân phối chiến dịch chăm sóc khách cũ với khoảng nghỉ tự nhiên nhằm đảm bảo độ ổn định kết nối cao nhất.

## 2. Luồng xử lý đơn hàng, khấu trừ tồn kho & tự động khớp chuyển khoản SePay

Khi nhân viên tư vấn chốt đơn ngay bên trong cửa sổ hội thoại Zalo trên ZaChat, hệ thống thực thi chuỗi tác vụ nguyên tử:
1. **Kiểm tra & giữ chỗ tồn kho (Inventory Reservation):** Tra cứu số lượng tồn kho khả dụng ngay cạnh khung chat; khi xác nhận tạo đơn, tồn kho được khấu trừ tức thì để ngăn chặn tình trạng bán vượt tồn giữa nhiều nhân viên trực cùng lúc.
2. **Ghi nhận Sổ công nợ (Receivables Ledger):** Cập nhật tổng giá trị đơn hàng, khoản thanh toán một phần và dư nợ lũy kế gắn trực tiếp với hồ sơ CRM của khách hàng.
3. **Đối soát Webhook SePay tự động:** Khi khách hàng quét mã chuyển khoản ngân hàng, sự kiện thanh toán từ SePay tự động đối chiếu số tiền và nội dung giao dịch để chuyển trạng thái đơn sang *Đã thanh toán* ngay trên màn hình làm việc.
